# ============================================================================
# SECTION 3: Text Preprocessing
# ============================================================================

This code prepares text data for an Image Captioning model. It cleans the raw descriptions and organizes them so each image is linked to its specific list of captions.

```python
def preprocess_caption(caption):
    caption = caption.lower()
    caption = re.sub(r'[^a-zA-Z0-9\s]', '', caption)
    caption = re.sub(r'\s+', ' ', caption).strip()
    caption = '<start> ' + caption + ' <end>'
    return caption

df['caption'] = df['caption'].apply(preprocess_caption)
image_to_captions = df.groupby('image')['caption'].apply(list).to_dict()

print(f"\nExample captions after preprocessing:")
for caption in list(image_to_captions.values())[0]:
    print(f"  {caption}")
```
